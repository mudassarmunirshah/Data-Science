#DATA VULIZATION

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go
from scipy import stats
import warnings
warnings.filterwarnings('ignore')

# Set matplotlib parameters for journal-quality figures (Elsevier/Springer standards)
plt.rcParams.update({
    'font.family': 'serif',
    'font.serif': ['Times New Roman'],
    'font.size': 10,
    'axes.labelsize': 10,
    'axes.titlesize': 12,
    'xtick.labelsize': 8,
    'ytick.labelsize': 8,
    'legend.fontsize': 8,
    'figure.dpi': 300,
    'savefig.dpi': 300,
    'savefig.format': 'pdf',
    'axes.grid': False,
    'axes.facecolor': 'white',
    'figure.facecolor': 'white'
})

# Color-blind safe palette
palette = sns.color_palette("colorblind")

def create_sample_dataframe():
    """Create a sample DataFrame matching the provided description."""
    np.random.seed(42)  # For reproducibility
    n = 100
    df = pd.DataFrame({
        'cement_kg_m3': np.random.randint(200, 500, n),
        'Rice_Husk_Ash_kg_per_m3': np.random.randint(0, 100, n),
        'mp_kg_m3': np.random.uniform(0, 200, n),
        'fa_kg_m3': np.random.uniform(500, 800, n),
        'ca_kg_m3': np.random.uniform(600, 1200, n),
        'water_kg_m3': np.random.uniform(150, 250, n),
        'superplasticizer_%Binder': np.random.uniform(0, 5, n),
        'compressive_strength_91d_MPa': np.random.uniform(20, 60, n)
    })
    return df

def preprocess_data(df):
    """Perform data preprocessing: report missing values and handle them."""
    # Report percentage of missing values per column
    missing_percent = (df.isnull().sum() / len(df)) * 100
    print("Percentage of missing values per column:")
    print(missing_percent)
    
    # Since all columns have 0% missing (as per description), no imputation needed.
    # If missing, apply mean imputation for numerical data (scientifically justified for small missing rates).
    # df.fillna(df.mean(), inplace=True)  # Uncomment if needed
    
    return df

def save_figure(fig, filename, interactive=False):
    """Save figure with meaningful filename."""
    if interactive:
        fig.write_html(f"{filename}.html")
    else:
        fig.savefig(f"{filename}.pdf", bbox_inches='tight')
        fig.savefig(f"{filename}.png", bbox_inches='tight')

def univariate_hist_kde(df, column):
    """Generate histogram + KDE for a numerical variable."""
    fig, ax = plt.subplots(figsize=(6, 4))
    sns.histplot(df[column], kde=True, ax=ax, color=palette[0])
    ax.set_xlabel(f"{column.replace('_', ' ').title()} ({get_unit(column)})")
    ax.set_ylabel("Frequency")
    ax.set_title(f"Distribution of {column.replace('_', ' ').title()}")
    save_figure(fig, f"univariate_hist_kde_{column}")
    plt.close()

def univariate_box_violin(df, column):
    """Generate boxplot and violin plot for outlier detection."""
    fig, axes = plt.subplots(1, 2, figsize=(10, 4))
    sns.boxplot(y=df[column], ax=axes[0], color=palette[1])
    axes[0].set_ylabel(f"{column.replace('_', ' ').title()} ({get_unit(column)})")
    axes[0].set_title("Boxplot")
    
    sns.violinplot(y=df[column], ax=axes[1], color=palette[2])
    axes[1].set_ylabel(f"{column.replace('_', ' ').title()} ({get_unit(column)})")
    axes[1].set_title("Violin Plot")
    
    save_figure(fig, f"univariate_box_violin_{column}")
    plt.close()

def bivariate_scatter_regression(df, x_col, y_col='compressive_strength_91d_MPa'):
    """Generate scatter plot with regression line."""
    fig, ax = plt.subplots(figsize=(6, 4))
    sns.regplot(x=df[x_col], y=df[y_col], ax=ax, scatter_kws={'alpha':0.6}, line_kws={'color':palette[3]})
    ax.set_xlabel(f"{x_col.replace('_', ' ').title()} ({get_unit(x_col)})")
    ax.set_ylabel(f"Compressive Strength (MPa)")
    ax.set_title(f"{x_col.replace('_', ' ').title()} vs Compressive Strength")
    save_figure(fig, f"bivariate_scatter_{x_col}_vs_strength")
    plt.close()

def multivariate_pairplot(df, vars_list):
    """Generate pairplot for key variables."""
    pairplot = sns.pairplot(df[vars_list], diag_kind='kde', plot_kws={'alpha':0.6})
    pairplot.fig.suptitle("Pairplot of Key Variables", y=1.02)
    save_figure(pairplot.fig, "multivariate_pairplot_key_vars")
    plt.close()

def correlation_heatmaps(df):
    """Generate Pearson and Spearman correlation heatmaps."""
    numerical_cols = df.select_dtypes(include=[np.number]).columns
    pearson_corr = df[numerical_cols].corr(method='pearson')
    spearman_corr = df[numerical_cols].corr(method='spearman')
    
    fig, axes = plt.subplots(1, 2, figsize=(12, 5))
    sns.heatmap(pearson_corr, annot=True, cmap='viridis', ax=axes[0], fmt='.2f', cbar=False)
    axes[0].set_title("Pearson Correlation")
    
    sns.heatmap(spearman_corr, annot=True, cmap='viridis', ax=axes[1], fmt='.2f', cbar=False)
    axes[1].set_title("Spearman Correlation")
    
    save_figure(fig, "correlation_heatmaps")
    plt.close()

def categorical_cement_bins(df):
    """Cement-bin-wise comparison of compressive strength using boxplot and violin plot."""
    # Bin cement into categories
    df['cement_bins'] = pd.cut(df['cement_kg_m3'], bins=3, labels=['Low', 'Medium', 'High'])
    
    fig, axes = plt.subplots(1, 2, figsize=(10, 4))
    sns.boxplot(x='cement_bins', y='compressive_strength_91d_MPa', data=df, ax=axes[0], palette=palette)
    axes[0].set_xlabel("Cement Bins (kg/m³)")
    axes[0].set_ylabel("Compressive Strength (MPa)")
    axes[0].set_title("Boxplot: Cement Bins vs Strength")
    
    sns.violinplot(x='cement_bins', y='compressive_strength_91d_MPa', data=df, ax=axes[1], palette=palette)
    axes[1].set_xlabel("Cement Bins (kg/m³)")
    axes[1].set_ylabel("Compressive Strength (MPa)")
    axes[1].set_title("Violin Plot: Cement Bins vs Strength")
    
    save_figure(fig, "categorical_cement_bins_vs_strength")
    plt.close()

def advanced_3d_scatter(df):
    """Generate 3D interactive scatter plot."""
    fig = px.scatter_3d(df, x='cement_kg_m3', y='water_kg_m3', z='compressive_strength_91d_MPa',
                        labels={'cement_kg_m3': 'Cement (kg/m³)', 'water_kg_m3': 'Water (kg/m³)', 'compressive_strength_91d_MPa': 'Strength (MPa)'},
                        title="3D Scatter: Cement, Water, and Strength")
    save_figure(fig, "advanced_3d_scatter", interactive=True)

def advanced_parallel_coordinates(df):
    """Generate parallel coordinates plot."""
    vars_list = ['cement_kg_m3', 'Rice_Husk_Ash_kg_per_m3', 'water_kg_m3', 'superplasticizer_%Binder', 'compressive_strength_91d_MPa']
    fig = px.parallel_coordinates(df[vars_list], color='compressive_strength_91d_MPa',
                                  labels={col: col.replace('_', ' ').title() + f" ({get_unit(col)})" for col in vars_list},
                                  title="Parallel Coordinates: Mix Constituents vs Strength")
    save_figure(fig, "advanced_parallel_coordinates", interactive=True)

def get_unit(column):
    """Helper to get units for columns."""
    units = {
        'cement_kg_m3': 'kg/m³',
        'Rice_Husk_Ash_kg_per_m3': 'kg/m³',
        'mp_kg_m3': 'kg/m³',
        'fa_kg_m3': 'kg/m³',
        'ca_kg_m3': 'kg/m³',
        'water_kg_m3': 'kg/m³',
        'superplasticizer_%Binder': '% Binder',
        'compressive_strength_91d_MPa': 'MPa'
    }
    return units.get(column, '')

def main():
    """Main function to run EDA and generate all figures."""
    df = create_sample_dataframe()
    df = preprocess_data(df)
    
    numerical_cols = df.select_dtypes(include=[np.number]).columns
    
    # A. Univariate analysis
    for col in numerical_cols:
        univariate_hist_kde(df, col)
        univariate_box_violin(df, col)
    
    # B. Bivariate & multivariate analysis
    input_cols = [col for col in numerical_cols if col != 'compressive_strength_91d_MPa']
    for col in input_cols:
        bivariate_scatter_regression(df, col)
    
    key_vars = ['cement_kg_m3', 'water_kg_m3', 'compressive_strength_91d_MPa']
    multivariate_pairplot(df, key_vars)
    correlation_heatmaps(df)
    
    # C. Categorical analysis
    categorical_cement_bins(df)
    
    # D. Advanced visualizations
    advanced_3d_scatter(df)
    advanced_parallel_coordinates(df)

if __name__ == "__main__":
    main()
```
