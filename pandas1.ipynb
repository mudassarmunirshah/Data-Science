```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go
from sklearn.impute import SimpleImputer
from scipy.stats import pearsonr, spearmanr
import warnings
warnings.filterwarnings('ignore')

# Set global plotting styles for journal quality
plt.rcParams.update({
    'font.family': 'serif',
    'font.serif': ['Times New Roman'],
    'font.size': 10,
    'axes.labelsize': 10,
    'axes.titlesize': 12,
    'xtick.labelsize': 8,
    'ytick.labelsize': 8,
    'legend.fontsize': 8,
    'figure.figsize': (8, 6),
    'savefig.dpi': 300,
    'savefig.format': 'pdf',
    'axes.grid': False,
    'axes.facecolor': 'white'
})
sns.set_palette("colorblind")  # Color-blind safe palette

# Create sample DataFrame based on description (for executability)
np.random.seed(42)
n = 100
df = pd.DataFrame({
    'cement_kg_m3': np.random.randint(200, 500, n),
    'rice_husk_ash_kg_m3': np.random.uniform(0, 50, n),
    'marbal_powder_kg_m3': np.random.uniform(0, 100, n),
    'fa_kg_m3': np.random.uniform(500, 800, n),
    'ca_kg_m3': np.random.uniform(1000, 1500, n),
    'water_kg_m3': np.random.uniform(150, 250, n),
    'superplasticizer_%Binder': np.random.uniform(0, 2, n),
    'compressive_strength_91d_MPa': np.random.uniform(20, 60, n),
    'cement_bins': pd.Categorical(np.random.choice(['Low', 'Medium', 'High'], n))
})

# Introduce missing values as per description
df.loc[np.random.choice(n, 4, replace=False), 'rice_husk_ash_kg_m3'] = np.nan
df.loc[np.random.choice(n, 1, replace=False), 'marbal_powder_kg_m3'] = np.nan
df.loc[np.random.choice(n, 2, replace=False), 'fa_kg_m3'] = np.nan
df.loc[np.random.choice(n, 6, replace=False), 'ca_kg_m3'] = np.nan
df.loc[np.random.choice(n, 1, replace=False), 'water_kg_m3'] = np.nan
df.loc[np.random.choice(n, 1, replace=False), 'superplasticizer_%Binder'] = np.nan

# Data preprocessing
def preprocess_data(df):
    # Report percentage of missing values per column
    missing_pct = df.isnull().sum() / len(df) * 100
    print("Percentage of missing values per column:")
    print(missing_pct)
    
    # Scientifically justified handling: Use mean imputation for continuous variables (assuming MAR and normal distribution)
    # For categorical, no imputation needed as it's complete
    num_cols = df.select_dtypes(include=[np.number]).columns
    imputer = SimpleImputer(strategy='mean')
    df[num_cols] = imputer.fit_transform(df[num_cols])
    
    return df

df = preprocess_data(df)

# Numerical columns for analysis
num_cols = ['cement_kg_m3', 'rice_husk_ash_kg_m3', 'marbal_powder_kg_m3', 'fa_kg_m3', 'ca_kg_m3', 'water_kg_m3', 'superplasticizer_%Binder', 'compressive_strength_91d_MPa']
key_vars = ['cement_kg_m3', 'water_kg_m3', 'compressive_strength_91d_MPa']  # Key variables for pairplot to avoid clutter

# Function to save figures
def save_figure(fig, filename, interactive=False):
    if interactive:
        fig.write_html(f"{filename}.html")
    else:
        fig.savefig(f"{filename}.pdf", bbox_inches='tight')
        fig.savefig(f"{filename}.png", bbox_inches='tight')

# A. Univariate analysis
def plot_univariate(df, col):
    fig, axes = plt.subplots(1, 3, figsize=(15, 5))
    
    # Histogram + KDE
    sns.histplot(df[col], kde=True, ax=axes[0])
    axes[0].set_title(f'Histogram + KDE of {col.replace("_", " ").title()}')
    axes[0].set_xlabel(f'{col.replace("_", " ").title()} ({get_unit(col)})')
    axes[0].set_ylabel('Frequency')
    
    # Boxplot
    sns.boxplot(y=df[col], ax=axes[1])
    axes[1].set_title(f'Boxplot of {col.replace("_", " ").title()}')
    axes[1].set_ylabel(f'{col.replace("_", " ").title()} ({get_unit(col)})')
    
    # Violin plot
    sns.violinplot(y=df[col], ax=axes[2])
    axes[2].set_title(f'Violin Plot of {col.replace("_", " ").title()}')
    axes[2].set_ylabel(f'{col.replace("_", " ").title()} ({get_unit(col)})')
    
    plt.tight_layout()
    save_figure(fig, f'univariate_{col}')
    plt.close()

def get_unit(col):
    units = {
        'cement_kg_m3': 'kg/m³',
        'rice_husk_ash_kg_m3': 'kg/m³',
        'marbal_powder_kg_m3': 'kg/m³',
        'fa_kg_m3': 'kg/m³',
        'ca_kg_m3': 'kg/m³',
        'water_kg_m3': 'kg/m³',
        'superplasticizer_%Binder': '% Binder',
        'compressive_strength_91d_MPa': 'MPa'
    }
    return units.get(col, '')

for col in num_cols:
    plot_univariate(df, col)

# B. Bivariate & multivariate analysis
def plot_scatter_regression(df, x_col, y_col='compressive_strength_91d_MPa'):
    fig, ax = plt.subplots(figsize=(8, 6))
    sns.regplot(x=x_col, y=y_col, data=df, ax=ax, scatter_kws={'alpha':0.6})
    ax.set_title(f'Scatter Plot: {x_col.replace("_", " ").title()} vs {y_col.replace("_", " ").title()}')
    ax.set_xlabel(f'{x_col.replace("_", " ").title()} ({get_unit(x_col)})')
    ax.set_ylabel(f'{y_col.replace("_", " ").title()} ({get_unit(y_col)})')
    plt.tight_layout()
    save_figure(fig, f'scatter_{x_col}_vs_{y_col}')
    plt.close()

for col in num_cols[:-1]:  # Exclude strength itself
    plot_scatter_regression(df, col)

# Pairplot for key variables
def plot_pairplot(df, vars_list):
    g = sns.pairplot(df[vars_list], diag_kind='kde', plot_kws={'alpha':0.6})
    g.fig.suptitle('Pairplot of Key Variables', y=1.02)
    for ax in g.axes.flatten():
        if ax:
            xlabel = ax.get_xlabel()
            ylabel = ax.get_ylabel()
            if xlabel in vars_list:
                ax.set_xlabel(f'{xlabel.replace("_", " ").title()} ({get_unit(xlabel)})')
            if ylabel in vars_list:
                ax.set_ylabel(f'{ylabel.replace("_", " ").title()} ({get_unit(ylabel)})')
    save_figure(g.fig, 'pairplot_key_vars')
    plt.close()

plot_pairplot(df, key_vars)

# Correlation heatmaps
def plot_correlation_heatmap(df, method='pearson'):
    corr = df[num_cols].corr(method=method)
    fig, ax = plt.subplots(figsize=(10, 8))
    mask = np.triu(np.ones_like(corr, dtype=bool))
    sns.heatmap(corr, mask=mask, annot=True, fmt='.2f', cmap='viridis', ax=ax, cbar_kws={'shrink': 0.8})
    ax.set_title(f'{method.capitalize()} Correlation Heatmap')
    save_figure(fig, f'correlation_heatmap_{method}')
    plt.close()

plot_correlation_heatmap(df, 'pearson')
plot_correlation_heatmap(df, 'spearman')

# C. Categorical analysis
def plot_categorical_comparison(df, cat_col='cement_bins', num_col='compressive_strength_91d_MPa'):
    fig, axes = plt.subplots(1, 2, figsize=(12, 6))
    
    # Boxplot
    sns.boxplot(x=cat_col, y=num_col, data=df, ax=axes[0])
    axes[0].set_title(f'Boxplot: {num_col.replace("_", " ").title()} by {cat_col.replace("_", " ").title()}')
    axes[0].set_xlabel(cat_col.replace("_", " ").title())
    axes[0].set_ylabel(f'{num_col.replace("_", " ").title()} ({get_unit(num_col)})')
    
    # Violin plot
    sns.violinplot(x=cat_col, y=num_col, data=df, ax=axes[1])
    axes[1].set_title(f'Violin Plot: {num_col.replace("_", " ").title()} by {cat_col.replace("_", " ").title()}')
    axes[1].set_xlabel(cat_col.replace("_", " ").title())
    axes[1].set_ylabel(f'{num_col.replace("_", " ").title()} ({get_unit(num_col)})')
    
    plt.tight_layout()
    save_figure(fig, f'categorical_{cat_col}_vs_{num_col}')
    plt.close()

plot_categorical_comparison(df)

# D. Advanced visualizations (Interactive with Plotly)
# 3D Scatter Plot
def plot_3d_scatter(df):
    fig = px.scatter_3d(df, x='cement_kg_m3', y='water_kg_m3', z='compressive_strength_91d_MPa',
                        title='3D Scatter Plot: Cement, Water, and Compressive Strength',
                        labels={'cement_kg_m3': 'Cement (kg/m³)', 'water_kg_m3': 'Water (kg/m³)', 'compressive_strength_91d_MPa': 'Strength (MPa)'})
    fig.update_layout(scene=dict(xaxis_title='Cement (kg/m³)', yaxis_title='Water (kg/m³)', zaxis_title='Strength (MPa)'))
    save_figure(fig, '3d_scatter_cement_water_strength', interactive=True)

plot_3d_scatter(df)

# Parallel Coordinates Plot
def plot_parallel_coordinates(df):
    vars_for_parallel = ['cement_kg_m3', 'rice_husk_ash_kg_m3', 'marbal_powder_kg_m3', 'fa_kg_m3', 'ca_kg_m3', 'water_kg_m3', 'superplasticizer_%Binder', 'compressive_strength_91d_MPa']
    fig = px.parallel_coordinates(df[vars_for_parallel], color='compressive_strength_91d_MPa',
                                  labels={col: f'{col.replace("_", " ").title()} ({get_unit(col)})' for col in vars_for_parallel},
                                  title='Parallel Coordinates Plot of Mix Constituents vs Strength')
    save_figure(fig, 'parallel_coordinates_mix_vs_strength', interactive=True)

plot_parallel_coordinates(df)
```
